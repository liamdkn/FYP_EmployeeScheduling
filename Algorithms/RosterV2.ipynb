```python

class Shift:
    shift_id_counter = 0

    def __init__(self, name, start_time, end_time, shift_map, assigned_employee):
        self.id = Shift.shift_id_counter
        Shift.shift_id_counter += 1
        self.name = name
        self.start_time = start_time
        self.end_time = end_time
        self.shift_map = shift_map
        self.assigned_employee = assigned_employee


shift1 = Shift(
    name="Morning Shift",
    start_time="08:00",
    end_time="12:00",
    shift_map={
        "Monday": True,
        "Tuesday": True,
        "Wednesday": True,
        "Thursday": True,
        "Friday": True,
        "Saturday": True,
        "Sunday": False,
    },
    assigned_employee=""
)

shift2 = Shift(
    name="Evening Shift",
    start_time="14:00",
    end_time="18:00",
    shift_map={
        "Monday": True,
        "Tuesday": True,
        "Wednesday": True,
        "Thursday": True,
        "Friday": True,
        "Saturday": True,
        "Sunday": True,
    },
    assigned_employee=""
)

shift3 = Shift(
    name="Lunch Shift",
    start_time="12:00",
    end_time="15:00",
    shift_map={
        "Monday": False,
        "Tuesday": False,
        "Wednesday": False,
        "Thursday": False,
        "Friday": True,
        "Saturday": True,
        "Sunday": True,
    },
    assigned_employee=""
)

shifts = [shift1, shift2, shift3]
```
