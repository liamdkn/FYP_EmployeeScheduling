```python
class Employee:
    def __init__(self, first_name, availability, days_per_week, movable):
        self.first_name = first_name
        self.availability = availability
        self.days_per_week = days_per_week
        self.movable = movable



employee1 = Employee(
    first_name="John",
    availability={
        "Monday": (time(8, 0), time(18, 0)),
        "Tuesday": (time(8, 0), time(18, 0)),
        "Wednesday": (time(8, 0), time(18, 0)),
        "Thursday": (time(8, 0), time(18, 0)),
        "Friday": (time(8, 0), time(18, 0)),
        "Saturday": (time(8, 0), time(18, 0)),
        "Sunday": (time(8, 0), time(18, 0)),
    },
    days_per_week = 5,
    movable=True
)

employee2 = Employee(
    first_name="Jane",
    availability={
        "Monday": (time(8, 0), time(18, 0)),
        "Tuesday": (time(8, 0), time(18, 0)),
        "Wednesday": (time(8, 0), time(18, 0)),
        "Thursday": (time(8, 0), time(18, 0)),
        "Friday": (time(8, 0), time(18, 0)),
        "Saturday": (time(8, 0), time(18, 0)),
        "Sunday": (time(8, 0), time(18, 0)),
    },
    days_per_week = 5,
    movable=True
)

employee3 = Employee(
    first_name="Joe",
    availability={
        "Monday": (time(8, 0), time(18, 0)),
        "Tuesday": (time(8, 0), time(18, 0)),
        "Wednesday": (time(8, 0), time(18, 0)),
        "Thursday": (time(8, 0), time(18, 0)),
        "Friday": (time(8, 0), time(18, 0)),
        "Saturday": (time(8, 0), time(18, 0)),
        "Sunday": (time(8, 0), time(18, 0)),
    },
    days_per_week = 5,
    movable=True
)

employees = [employee1, employee2, employee3]
```

```python
class Ind_Shift:
    shift_id_counter = 0

    def __init__(self, name, start_time, end_time, day, assigned_employee):
        self.id = Ind_Shift.shift_id_counter
        Ind_Shift.shift_id_counter += 1
        self.name = name
        self.day = day
        self.start_time = start_time
        self.end_time = end_time
        self.assigned_employee = assigned_employee
        

class Shifts:
    def __init__(self, name, start_time, end_time, shift_map):
        self.name = name
        self.start_time = start_time
        self.end_time = end_time
        self.shift_map = shift_map
        

shift1 = Shifts(
    name="Morning Shift",
    start_time= time(8, 0),
    end_time= time(12, 0),
    shift_map={
        "Monday": True,
        "Tuesday": True,
        "Wednesday": True,
        "Thursday": True,
        "Friday": True,
        "Saturday": True,
        "Sunday": True,
    },
)

shift2 = Shifts(
    name="Evening Shift",
    start_time= time(14, 0),
    end_time= time(18, 0),
    shift_map={
        "Monday": True,
        "Tuesday": True,
        "Wednesday": True,
        "Thursday": True,
        "Friday": True,
        "Saturday": True,
        "Sunday": True,
    },
)

shift3 = Shifts(
    name="Lunch Shift",
    start_time= time(12, 0),
    end_time= time(15, 0),
    shift_map={
        "Monday": False,
        "Tuesday": False,
        "Wednesday": False,
        "Thursday": False,
        "Friday": False,
        "Saturday": True,
        "Sunday": False,
    },
)

shifts = [shift1, shift2, shift3]
```

```python
class DayTimeOffRequest:
    def __init__(self, name, day, start_time, end_time):
        self.employee_name = name
        self.day = day
        self.start_time = start_time
        self.end_time = end_time


# Example DayTimeOffRequest objects
day_time_off_requests = [
    DayTimeOffRequest(name="John", day= days[1], start_time=time(9, 0), end_time=time(12, 0)),
    DayTimeOffRequest(name="Jane", day= days[4], start_time=None, end_time=None),  # Full day off
    DayTimeOffRequest(name="Joe", day= days[6], start_time=time(14, 0), end_time=time(17, 0)),
]
```

Create an object for every shift

```python
individual_shifts = []

for shift in shifts:
    for day, is_shift_day in shift.shift_map.items():
        if is_shift_day:
            individual_shift = Ind_Shift(
                name= shift.name,
                day = day,
                start_time=shift.start_time,
                end_time=shift.end_time,
                assigned_employee = None
            )
            individual_shifts.append(individual_shift)

# Print the result
for shift in individual_shifts:
    print(f"Shift ID: {shift.id}, Day: {shift.day}, Name: {shift.name}, Start Time: {shift.start_time}, End Time: {shift.end_time}, Assigned Employee: {shift.assigned_employee}")
```

Assign a random employee to every shift 

```python
def is_employee_available(employee, shift):
    if shift.day in employee.availability:
        shift_start_time = shift.start_time
        shift_end_time = shift.end_time

        # Get the employee's availability for the day
        employee_start_time, employee_end_time = employee.availability[shift.day]

        # Check if the employee can work during the shift hours
        if (
            employee_start_time <= shift_start_time <= employee_end_time
            and employee_start_time <= shift_end_time <= employee_end_time
        ):
            
            return True

    return False
```

```python
def initialize_population():
    assigned_shifts_count = {employee: 0 for employee in employees}

    for shift in individual_shifts:
        available_employees = [employee for employee in employees if is_employee_available(employee, shift)]

        if available_employees:
            random_employee = random.choice(available_employees)
            shift.assigned_employee = random_employee.first_name
            assigned_shifts_count[random_employee] += 1
        else:
            shift.assigned_employee = None  # No available employees for this shift

    return individual_shifts, assigned_shifts_count

# Example usage
updated_individual_shifts, assigned_shifts_count = initialize_population()

# Print the assigned shifts
for shift in updated_individual_shifts:
    print(f"ID: {shift.id}, {shift.day} {shift.name}, Start Time: {shift.start_time}, End Time: {shift.end_time}, Assigned Employee: {shift.assigned_employee}")

# Print the total number of shifts and how many each employee got
print("\nShifts Assignment Summary:")
print(len(individual_shifts), " Shifts to be assigned")
total_shifts = len(updated_individual_shifts)
for employee, shifts_count in assigned_shifts_count.items():
    print(f"{employee.first_name}: Assigned {shifts_count} out of {total_shifts} shifts")
```

# Fitness Checking

```python
def calculate_fitness():
    # Is every shift assigned 
    total_shifts = len(individual_shifts)
    covered_shifts = sum(1 for shift in individual_shifts if shift.assigned_employee)
    coverage_fitness = (covered_shifts / total_shifts)  * 100

    # Are employees available
    availability_fitness = 0
    for shift in individual_shifts:
        assigned_employee = shift.assigned_employee
        if assigned_employee:
            employee = next((e for e in employees if e.first_name == assigned_employee), None)
            if employee and shift.day in employee.availability:
                start_time, end_time = employee.availability[shift.day]
                if shift.start_time >= start_time and shift.end_time <= end_time:
                    availability_fitness += 1

    availability_fitness = availability_fitness / len(individual_shifts) * 100

    # Max Consecutively days
    max_consecutive_days_allowed = 5

    def calculate_consecutive_days(employee_schedule):
        consecutive_days = 0
        max_consecutive_days = 0
        for shift in employee_schedule:
            if shift.assigned_employee:
                consecutive_days += 1
            else:
                max_consecutive_days = max(max_consecutive_days, consecutive_days)
                consecutive_days = 0

        return max(max_consecutive_days, consecutive_days)

    total_employees = len(employees)
    violating_employees = 0

    for employee in employees:
        employee_schedule = [shift for shift in individual_shifts if shift.assigned_employee == employee.first_name]
        consecutive_days = calculate_consecutive_days(employee_schedule)

        if consecutive_days > max_consecutive_days_allowed:
            violating_employees += 1

    percentage_fitness = (violating_employees / total_employees) * 100

    # Employee requested days off
    day_off_fitness = 0

    for day_off_request in day_time_off_requests:
        employee_name = day_off_request.employee_name
        day = day_off_request.day
        start_time = day_off_request.start_time
        end_time = day_off_request.end_time

        for shift in individual_shifts:
            if shift.assigned_employee == employee_name and shift.day == day:
                if start_time is not None and end_time is not None:
                    # Check for time range
                    if shift.start_time >= start_time and shift.end_time <= end_time:
                        # Honor the day-off request
                        day_off_fitness += 1  # Increment the fitness for honoring the day-off request
                else:
                    # Honor the day-off request (full day off)
                    day_off_fitness += 1  # Increment the fitness for honoring the day-off request

    day_off_fitness_percentage = (day_off_fitness / len(day_time_off_requests)) * 100

    # Employee only work 1 shift a day
    one_shift_per_day_fitness = 0

    for day in days:
        shifts_for_day = [shift for shift in individual_shifts if shift.day == day and shift.assigned_employee]

        employees_working_on_day = set()
        for shift in shifts_for_day:
            if shift.assigned_employee in employees_working_on_day:
                # Employee is already assigned a shift for this day
                one_shift_per_day_fitness += 1  # Increment the fitness for violating the constraint
            else:
                employees_working_on_day.add(shift.assigned_employee)

    one_shift_per_day_fitness_percentage = (one_shift_per_day_fitness / len(individual_shifts)) * 100


    # Creating a dictionary with fitness values
    fitness_data = {
        'Criterion': ['Shift Coverage', 'Availability', 'Consecutive Days', 'Day Off Requests', 'One  Shift a Day'],
        'Fitness (%)': [coverage_fitness, availability_fitness, percentage_fitness, day_off_fitness_percentage, one_shift_per_day_fitness]
    }

    # Creating a DataFrame from the dictionary
    fitness_df = pd.DataFrame(fitness_data)
    
    return coverage_fitness, availability_fitness, percentage_fitness, day_off_fitness_percentage, one_shift_per_day_fitness_percentage

```

```python
best_population = None
best_fitness = float('-inf')

for _ in range(100):
    current_population, _ = initialize_population()
    coverage_fitness, availability_fitness, percentage_fitness, day_off_fitness_percentage, one_shift_per_day_fitness_percentage = calculate_fitness()

    # Calculate the cumulative total fitness out of 100
    cumulative_total_fitness = (coverage_fitness + availability_fitness + percentage_fitness + day_off_fitness_percentage + one_shift_per_day_fitness_percentage) / 5

    # Check if the current population is the best so far
    if cumulative_total_fitness > best_fitness:
        best_fitness = cumulative_total_fitness
        best_population = current_population

    # Print the breakdown of fitness for the current population
    

# Print or use the best population and its fitness
print("Best Total Fitness", best_fitness)
print(f"Coverage Fitness: {coverage_fitness}")
print(f"Availability Fitness: {availability_fitness}")
print(f"Consecutive Days Fitness: {percentage_fitness}")
print(f"Day Off Fitness: {day_off_fitness_percentage}")
print(f"One shift per day Fitness:  {one_shift_per_day_fitness_percentage}")
print("----")
```
