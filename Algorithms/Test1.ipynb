```python
import random
from datetime import time
from datetime import datetime


class Employee:
    def __init__(self, first_name, availability):
        self.first_name = first_name
        self.availability = availability

class Shift:
    def __init__(self, name, start_time, end_time, shift_map):
        self.name = name
        self.start_time = start_time
        self.end_time = end_time
        self.shift_map = shift_map

# Example employees and shifts
employee1 = Employee(
    first_name="John",
    availability={
        "Monday": (time(8, 0), time(18, 0)),
        "Wednesday": (time(8, 0), time(18, 0)),
        "Thursday": (time(8, 0), time(18, 0)),
        "Friday": (time(8, 0), time(18, 0)),
        "Saturday": (time(8, 0), time(18, 0)),
        "Sunday": (time(8, 0), time(18, 0)),
    }
)

employee2 = Employee(
    first_name="Jane",
    availability={
        "Monday": (time(8, 0), time(18, 0)),
        "Tuesday": (time(8, 0), time(18, 0)),
        "Wednesday": (time(8, 0), time(18, 0)),
        "Thursday": (time(8, 0), time(18, 0)),
        "Friday": (time(8, 0), time(18, 0)),
        "Saturday": (time(8, 0), time(18, 0)),
    }
)

employees = [employee1, employee2]

shift1 = Shift(
    name="Morning Shift",
    start_time="08:00",
    end_time="12:00",
    shift_map={
        "Monday": True,
        "Tuesday": True,
        "Wednesday": True,
        "Thursday": True,
        "Friday": True,
        "Saturday": True,
        "Sunday": False,
    }
)

shift2 = Shift(
    name="Evening Shift",
    start_time="14:00",
    end_time="18:00",
    shift_map={
        "Monday": True,
        "Tuesday": True,
        "Wednesday": True,
        "Thursday": True,
        "Friday": True,
        "Saturday": True,
        "Sunday": True,
    }
)

shifts = [shift1, shift2]

# Genetic Algorithm Parameters
population_size = 10
generations = 100

def generate_individual():
    return [random.choice(employees) for _ in shifts]

def calculate_fitness(individual):
    fitness = 0
    for shift, employee in zip(shifts, individual):
        for day, availability in employee.availability.items():
            shift_start_time = datetime.strptime(shift.start_time, "%H:%M").time()
            shift_end_time = datetime.strptime(shift.end_time, "%H:%M").time()
            employee_start_time = availability[0]
            employee_end_time = availability[1]
            if shift.shift_map[day] and employee_start_time <= shift_start_time <= employee_end_time:
                fitness += 1
    return fitness

def crossover(parent1, parent2):
    # One-point crossover
    crossover_point = random.randint(0, len(shifts) - 1)
    child1 = parent1[:crossover_point] + parent2[crossover_point:]
    child2 = parent2[:crossover_point] + parent1[crossover_point:]
    return child1, child2

def mutate(individual):
    # Randomly replace an employee in the individual
    mutated_individual = individual.copy()
    mutation_point = random.randint(0, len(shifts) - 1)
    mutated_individual[mutation_point] = random.choice(employees)
    return mutated_individual

# Genetic Algorithm
population = [generate_individual() for _ in range(population_size)]

for generation in range(generations):
    # Evaluate fitness of each individual
    fitness_scores = [calculate_fitness(individual) for individual in population]

    # Select parents for crossover
    parent1 = random.choice(population)
    parent2 = random.choice(population)

    # Apply crossover to create new individuals
    child1, child2 = crossover(parent1, parent2)

    # Apply mutation to new individuals
    child1 = mutate(child1)
    child2 = mutate(child2)

    # Replace two least fit individuals with the new children
    population[fitness_scores.index(min(fitness_scores))] = child1
    population[fitness_scores.index(min(fitness_scores))] = child2

# Select the best solution from the final population
best_solution = max(population, key=calculate_fitness)

# Display the best solution
print("Best Solution:")
for shift, employee in zip(shifts, best_solution):
    print(f"{employee.first_name} is assigned to {shift.name}")
```
