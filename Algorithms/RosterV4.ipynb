```python
class Ind_Shift:
    shift_id_counter = 1

    def __init__(self, name, start_time, end_time, day, employee):
        self.id = Ind_Shift.shift_id_counter
        Ind_Shift.shift_id_counter += 1
        self.name = name
        self.day = day
        self.start_time = start_time
        self.end_time = end_time
        self.employee = employee
        

class Shifts:
    def __init__(self, name, start_time, end_time, shift_map):
        self.name = name
        self.start_time = start_time
        self.end_time = end_time
        self.shift_map = shift_map
        

shift1 = Shifts(
    name="Morning Shift",
    start_time= time(8, 0),
    end_time= time(12, 0),
    shift_map={
        "Monday": True,
        "Tuesday": True,
        "Wednsday": True,
        "Thursday": True,
        "Friday": True,
        "Saturday": True,
        "Sunday": True,
    },
)

shift2 = Shifts(
    name="Evening Shift",
    start_time= time(14, 0),
    end_time= time(18, 0),
    shift_map={
        "Monday": True,
        "Tuesday": True,
        "Wednsday": True,
        "Thursday": True,
        "Friday": True,
        "Saturday": True,
        "Sunday": True,
    },
)

shift3 = Shifts(
    name="Lunch Shift",
    start_time= time(12, 0),
    end_time= time(15, 0),
    shift_map={
        "Monday": False,
        "Tuesday": False,
        "Wednsday": False,
        "Thursday": False,
        "Friday": False,
        "Saturday": True,
        "Sunday": False,
    },
)

shifts = [shift1, shift2, shift3]
```

```python
individual_shifts = []

for shift in shifts:
    for day, is_shift_day in shift.shift_map.items():
        if is_shift_day:
            individual_shift = Ind_Shift(
                name= shift.name,
                day = day,
                start_time=shift.start_time,
                end_time=shift.end_time,
                employee = None
            )
            individual_shifts.append(individual_shift)

for shift in individual_shifts:
    print(f"Shift ID: {shift.id}, Day: {shift.day}, Name: {shift.name}, Start Time: {shift.start_time}, End Time: {shift.end_time}, Assigned Employee: {shift.employee}")
```

```python
# Constants
num_employees = len(employees)
num_shifts = len(individual_shifts)

# Generate a random chromosome (assignment of shifts to employees)
def generate_random_chromosome():
    return [random.randint(1, num_employees) for _ in range(num_shifts)]

# Generate an initial population
def initialize_population(population_size):
    return [generate_random_chromosome() for _ in range(population_size)]

# Example usage
population_size = 10
initial_population = initialize_population(population_size)

# Print the initial population
for chromosome in initial_population:
    print(chromosome)

```

```python
def is_employee_available(employee, shift):
    if shift.day in employee.availability:
        shift_start_time = shift.start_time
        shift_end_time = shift.end_time

        employee_start_time, employee_end_time = employee.availability[shift.day]

        if (
            employee_start_time <= shift_start_time <= employee_end_time
            and employee_start_time <= shift_end_time <= employee_end_time
        ):
            
            return True

    return False
```

```python
def calculate_fitness(chromosome, employees, individual_shifts):
    total_shifts = len(chromosome)
    shifts_per_day = {}
    numberOfConflicts = 0
    max_shifts_per_week = 5

    # Count the number of shifts assigned to each employee
    shifts_per_employee = {employee.id: 0 for employee in employees}

    for i, employee_id in enumerate(chromosome):
            shift = individual_shifts[i]
            day = shift.day
            employee = employees[employee_id - 1]

            # Check if shift is within employees availability 
            if not (is_employee_available(employee, shift)):
                numberOfConflicts += 1

            # Check if an employee is assigned to more than one shift per day
            if day in shifts_per_day:
                if employee_id in shifts_per_day[day]:
                    numberOfConflicts += 1
                else:
                    shifts_per_day[day].add(employee_id)
            else:
                shifts_per_day[day] = {employee_id}

            # Count the number of shifts assigned to each employee
            shifts_per_employee[employee.id] += 1

    #Check if an employee is working more than 5 days
    shifts_penalty = sum(max(0, count - max_shifts_per_week) for count in shifts_per_employee.values())
    numberOfConflicts += shifts_penalty
    
    fitness = 1 / ((1.0 * numberOfConflicts + 1))
    
    return fitness, numberOfConflicts

for chromosome in initial_population:
    fitness, conflicts = calculate_fitness(chromosome, employees, individual_shifts)
    
    # Convert employee IDs in the chromosome to employee names
    employee_id = [employees[employee_id - 1].id if employee_id != 0 else "Unassigned" for employee_id in chromosome]

    print(f"Chromosome: {employee_id}, Fitness: {fitness}, Conflicts: {conflicts}")
```

```python
import random

def roulette_wheel_selection(population, num_selection):
    total_fitness = sum(individual['fitness'] for individual in population)
    selection_probabilities = [individual['fitness'] / total_fitness for individual in population]
    selected_indices = random.choices(range(len(population)), weights=selection_probabilities, k=num_selection)
    selected_individuals = [population[i] for i in selected_indices]
    return selected_individuals

# Example usage:
num_selection = 2  # Adjust as needed
selected_for_crossover = roulette_wheel_selection(initial_population_with_fitness, num_selection)

for selected_individual in selected_for_crossover:
    print(f"Selected: Chromosome: {selected_individual['chromosome']}, Fitness: {selected_individual['fitness']}, Conflicts: {selected_individual['conflicts']}")
```

```python
def crossover(parent1, parent2):
    # Implement your crossover logic based on your chromosome representation
    crossover_point = random.randint(1, len(parent1) - 1)
    child1 = parent1[:crossover_point] + parent2[crossover_point:]
    child2 = parent2[:crossover_point] + parent1[crossover_point:]
    return child1, child2

# Example usage:
offspring = []

# Perform crossover on pairs of selected individuals
for i in range(0, len(selected_for_crossover), 2):
    parent1 = selected_for_crossover[i]['chromosome']
    parent2 = selected_for_crossover[i + 1]['chromosome']
    
    child1, child2 = crossover(parent1, parent2)
    
    # Note: You may choose to apply mutation to the offspring here if needed

    child1_conflicts, child1_fitness = calculate_fitness(child1, employees, individual_shifts)
    child2_conflicts, child2_fitness = calculate_fitness(child2, employees, individual_shifts)
    
    offspring.append({f"chromosome': {child1}, 'fitness': {child1_fitness}, 'conflicts': {child1_conflicts}"})
    offspring.append({f"chromosome': {child2}, 'fitness': {child2_fitness}, 'conflicts': {child2_conflicts}"})

# Now, 'offspring' contains the result of crossover
print(offspring)
```

```python
def print_schedule(chromosome, individual_shifts, employees):
    schedule = {}

    for i, employee_id in enumerate(chromosome):
        if employee_id != 0:  # Assuming 0 represents an unassigned shift
            shift = individual_shifts[i]
            employee = employees[employee_id - 1]  # Adjusting index for list access

            if employee_id not in schedule:
                schedule[employee_id] = []

            schedule[employee_id].append({
                "shift_id": shift.id,
                "day": shift.day,
                "start_time": shift.start_time,
                "end_time": shift.end_time,
                "employee_name": employee.first_name
            })

    # Print the schedule
    for employee_id, shifts in schedule.items():
        print(f"Employee {employee_id} ({employees[employee_id - 1].first_name}):")
        for shift in shifts:
            print(f"  Shift {shift['shift_id']} - {shift['day']} - {shift['start_time']} to {shift['end_time']}")
        print("\n")

# Example usage
chromosome = [1, 1, 2, 1, 3, 1, 2, 3, 2, 3, 2, 3, 2, 1, 3]
print_schedule(chromosome, individual_shifts, employees)
```

```python
import pandas as pd

def create_schedule_dataframe(chromosome, individual_shifts, employees):
    schedule = {}

    for i, employee_id in enumerate(chromosome):
        if employee_id != 0:  # Assuming 0 represents an unassigned shift
            shift = individual_shifts[i]
            employee = employees[employee_id - 1]  # Adjusting index for list access

            if employee_id not in schedule:
                schedule[employee_id] = {}

            schedule[employee_id][shift.day] = f"{shift.start_time} - {shift.end_time}"

    # Create a DataFrame
    df_schedule = pd.DataFrame(columns=["Monday", "Tuesday", "Wednsday", "Thursday", "Friday", "Saturday", "Sunday"])

    # Populate the DataFrame
    for employee_id, shifts in schedule.items():
        row_data = {day: shifts.get(day, "") for day in df_schedule.columns}
        df_schedule = df_schedule.append(pd.Series(row_data, name=f"Employee {employee_id}"))

    return df_schedule

# Example usage
chromosome = [3, 2, 3, 1, 3, 1, 2, 3, 2, 3, 2, 3, 2, 1, 3]
df_schedule = create_schedule_dataframe(chromosome, individual_shifts, employees)
df_schedule
```
